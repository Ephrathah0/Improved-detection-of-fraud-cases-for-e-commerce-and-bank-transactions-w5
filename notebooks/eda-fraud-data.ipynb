```python
import socket
import struct
```

```python
ip_df= pd.read_csv('../data/raw/IpAddress_to_Country.csv')
# Clean column names
ip_df.columns = ip_df.columns.str.strip().str.lower()  # all lowercase and stripped

# Check columns
print(ip_df.columns.tolist())  # should show ['lower_bound_ip_address', 'upper_bound_ip_address', 'country']

# Function to convert IP string to integer
def ip_to_int(ip):
    if pd.isna(ip):
        return None
    try:
        return struct.unpack("!I", socket.inet_aton(str(ip)))[0]
    except:
        return None
```

```python
ip_df['lower_bound_ip_int'] = ip_df['lower_bound_ip_address'].apply(ip_to_int)
ip_df['upper_bound_ip_int'] = ip_df['upper_bound_ip_address'].apply(ip_to_int)
```

```python
df['ip_int'] = df['ip_address'].apply(ip_to_int)

df['country'] = None

for idx, row in ip_df.iterrows():
    mask = (df['ip_int'] >= row['lower_bound_ip_int']) & (df['ip_int'] <= row['upper_bound_ip_int'])
    df.loc[mask, 'country'] = row['country']
```

```python
import pandas as pd
```

```python
df= pd.read_csv('../data/raw/Fraud_Data.csv')
```

```python
df['transaction_count'] = df.groupby('user_id')['user_id'].transform('count')
```

```python
df['purchase_time'] = pd.to_datetime(df['purchase_time'], errors='coerce')
```

```python
df['day_of_week'] = df['purchase_time'].dt.dayofweek 
```

```python
df['purchase_time'] = pd.to_datetime(df['purchase_time'], errors='coerce')
df['signup_time'] = pd.to_datetime(df['signup_time'], errors='coerce')
df['time_since_signup'] = (
    df['purchase_time'] - df['signup_time']
).dt.total_seconds() / 3600
```

```python
df['avg_purchase_value'] = df.groupby('user_id')['purchase_value'].transform('mean')
```

```python
df['signup_time'] = pd.to_datetime(df['signup_time'], errors='coerce')
df['purchase_time'] = pd.to_datetime(df['purchase_time'], errors='coerce')

df = df.dropna(subset=['signup_time', 'purchase_time'])

```

```python
df['time_since_signup'] = (
    df['purchase_time'] - df['signup_time']
).dt.total_seconds() / 3600  # hours
```

```python
df['txn_count_user'] = df.groupby('user_id')['user_id'].transform('count')
```

```python
df_model = df.drop(columns=[
    'user_id',
    'device_id',
    'ip_address',
    'ip_int',
    'lower_bound_ip_address',
    'upper_bound_ip_address'
], errors='ignore')
```
